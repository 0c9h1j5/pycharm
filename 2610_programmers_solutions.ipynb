```
#마지막 두 원소

def solution(num_list):
    if num_list[-1] > num_list[-2]:
        num_list.append(num_list[-1] - num_list[-2])

    else:
        num_list.append(num_list[-1]*2)

    return num_list


solution([2, 1, 6])
```

```
#이어 붙인 수

def solution(num_list):
    num1 = ''
    num2 = ''
    for i in num_list:
        if i%2 != 0:
            num1 += str(i)
        else:
            num2 += str(i)
    return int(num1) + int(num2)

solution([3,4,5,2,1])
```

```
#특정한 문자를 대문자로 바꾸기

def solution(my_string, alp):
    answer = ''
    if alp in my_string:
        for i in my_string:
            if i == alp:
                answer += i.upper()
                continue
            answer += i
    else:
        return my_string

    return answer
solution('programmers', 'p')
```

```
#꼬리 문자열

def solution(str_list, ex):
    answer = ''

    for i in str_list:
        if ex in i:
            continue
        else:
            answer += i
    return answer

solution(["abc", "bbc", "cbc"],"c")
```

```
#길이에 따른 연산

def solution(num_list):
    answer = 1
    if len(num_list) < 11:
        for n in num_list:
            answer *= n
    else:
        return sum(num_list)

    return answer

num_list = [2, 3, 4, 5]
solution(num_list)
```

```
# 뒤에서 5등 뒤로

def solution(num_list):
    answer = sorted(num_list)
    return answer[5:]

solution([12, 4, 15, 46, 38, 1, 14, 56, 32, 10])
```

```
#접미사인지 확인하기

def solution(my_string, is_suffix):
    if my_string[-len(is_suffix):]==is_suffix:
        return 1
    else:
        return 0



solution("banana","ana")
```

```
#원하는 문자열 찾기

def solution(myString, pat):
    lower1 = myString.lower()
    lower2 = pat.lower()

    if lower2 in lower1:
        return 1
    else:
        return 0

solution("AbCdEfG","aBc")
```

```
# 접두사인지 확인하기

def solution(my_string, is_prefix):
    if my_string[:len(is_prefix)] == is_prefix:
        return 1
    else:
        return 0

solution("banana", 'ban')
```

```
# 문자열 바꿔서 찾기

def solution(myString, pat):
    answer = ''
    for i in myString:
        if i == 'A':
            answer += 'B'
        else:
            answer += 'A'

    if pat in answer:
        return 1
    else:
        return 0

solution("ABBAA", "AABB")
```

```
# A강조하기

def solution(myString):
    myString = myString.lower()
    answer =''
    for i in myString:
        if i == 'a':
            answer += 'A'
            continue
        answer += i
    return answer


solution("abstract algebra")
```

```
# 정수 찾기
def solution(num_list, n):
    if n in num_list:
        return 1
    else:
        return 0


solution([1,2,3,4,5], 3)
```

```
# 배열의 원소 삭제하기
def solution(arr, delete_list):
    answer = []
    for i in arr:
        if i in delete_list:
            continue
        else:
            answer.append(i)
    return answer


solution([293, 1000, 395, 678, 94],[94, 777, 104, 1000, 1, 12])
```
