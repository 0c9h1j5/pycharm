```
#마지막 두 원소

def solution(num_list):
    if num_list[-1] > num_list[-2]:
        num_list.append(num_list[-1] - num_list[-2])

    else:
        num_list.append(num_list[-1]*2)

    return num_list


solution([2, 1, 6])
```

```
#이어 붙인 수

def solution(num_list):
    num1 = ''
    num2 = ''
    for i in num_list:
        if i%2 != 0:
            num1 += str(i)
        else:
            num2 += str(i)
    return int(num1) + int(num2)

solution([3,4,5,2,1])
```

```
#특정한 문자를 대문자로 바꾸기

def solution(my_string, alp):
    answer = ''
    if alp in my_string:
        for i in my_string:
            if i == alp:
                answer += i.upper()
                continue
            answer += i
    else:
        return my_string

    return answer
solution('programmers', 'p')
```

```
#꼬리 문자열

def solution(str_list, ex):
    answer = ''

    for i in str_list:
        if ex in i:
            continue
        else:
            answer += i
    return answer

solution(["abc", "bbc", "cbc"],"c")
```

```
#길이에 따른 연산

def solution(num_list):
    answer = 1
    if len(num_list) < 11:
        for n in num_list:
            answer *= n
    else:
        return sum(num_list)

    return answer

num_list = [2, 3, 4, 5]
solution(num_list)
```

```
# 문자열 바꿔서 찾기

def solution(myString, pat):
    answer = ''
    for i in myString:
        if i == 'A':
            answer += 'B'
        else:
            answer += 'A'

    if pat in answer:
        return 1
    else:
        return 0

solution("ABBAA", "AABB")
```

```
# 0 떼기

def solution(n_str):
    count = 0
    for i in range(len(n_str)):
        if n_str[i] == '0':
            count += 1
        else:
            break


    return n_str[count:]


solution('854020')
```

```
# 배열 비교하기

def solution(arr1, arr2):
    if len(arr1) < len(arr2):
        return -1
    elif len(arr1) > len(arr2):
        return 1
    else:
        if sum(arr1) > sum(arr2):
            return 1
        elif sum(arr1) < sum(arr2):
            return -1
        else:
            return 0


solution([100, 17, 84, 1],[55, 12, 65, 36])
```

```
#암호 해독

def solution(cipher, code):
    answer = ''
    for i in range(len(cipher)):
        if i%code == 0 and i > 0:
            answer += cipher[i-1]

    if len(answer) != len(cipher)//code:
        answer += cipher[-1]

    return answer

# chiper[code-1::code] 한줄이면 끝날껄..

solution("dfjardstddetckdaccccdegk",4)
```

```
#배열 회전시키기

def solution(numbers, direction):
    answer = []
    if direction == 'right':
        answer += numbers[-1:]
        answer += numbers[:-1]
    else:
        answer += numbers[1:]
        answer += numbers[:1]
    return answer

solution([4, 455, 6, 4, -1, 45, 6], 'left')
```
