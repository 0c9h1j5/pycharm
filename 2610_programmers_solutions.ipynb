```
#마지막 두 원소

def solution(num_list):
    if num_list[-1] > num_list[-2]:
        num_list.append(num_list[-1] - num_list[-2])

    else:
        num_list.append(num_list[-1]*2)

    return num_list


solution([2, 1, 6])
```

```
#이어 붙인 수

def solution(num_list):
    num1 = ''
    num2 = ''
    for i in num_list:
        if i%2 != 0:
            num1 += str(i)
        else:
            num2 += str(i)
    return int(num1) + int(num2)

solution([3,4,5,2,1])
```

```
#특정한 문자를 대문자로 바꾸기

def solution(my_string, alp):
    answer = ''
    if alp in my_string:
        for i in my_string:
            if i == alp:
                answer += i.upper()
                continue
            answer += i
    else:
        return my_string

    return answer
solution('programmers', 'p')
```

```
#꼬리 문자열

def solution(str_list, ex):
    answer = ''

    for i in str_list:
        if ex in i:
            continue
        else:
            answer += i
    return answer

solution(["abc", "bbc", "cbc"],"c")
```

```
#길이에 따른 연산

def solution(num_list):
    answer = 1
    if len(num_list) < 11:
        for n in num_list:
            answer *= n
    else:
        return sum(num_list)

    return answer

num_list = [2, 3, 4, 5]
solution(num_list)
```

```
# 뒤에서 5등 뒤로

def solution(num_list):
    answer = sorted(num_list)
    return answer[5:]

solution([12, 4, 15, 46, 38, 1, 14, 56, 32, 10])
```
